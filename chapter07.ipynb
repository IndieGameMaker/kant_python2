## 함수 (Function)

- 반복되는 로직을 합쳐놓은 문법
- 재사용 높일수 있는 코드 기법
- 단독사용가능 print(), len(), ...
- 함수명 컨벤션 : snake_case
- 매개변수 (Parameter)

```python
def 함수이름(매개변수, 매개변수, ...):
    실행할 로직 #1
    실행할 로직 #2
    실행할 로직 #3
```

````python
# 함수 정의
def say_hello(name, age=25, point=1000):
    print(f"안녕하세요. 저는 {name} 입니다. 나이는 {age}입니다.")
    print(f"초기 지급 포인트 {point}점이 지급되었습니다.")

say_hello("이재현")
````

````python
# 가변 인수
# *args

def add_all(*numbers): # 매개변수 (Parameter)
    print(f"입력된 튜플: {numbers}")
    total = sum(numbers)
    print(f"합계 결과: {total}")

add_all(1, 2, 3) # 인수 (argument)
add_all(1, 2, 3, 4, 5)
add_all()
````

````python
# 딕셔너리 가변 인수
# **kwargs
def user_info(**data):
    print(f"전달된 딕셔너리 구조: {data}")
    for key, value in data.items():
        print(f"{key}: {value}")

# 함수 호출
user_info(name="스파르타", level=5, active=True)
````

````python
# return - 결괏값 되돌리기
# 하나의 반환값

def check_ord_number(number):
    if number > 0:
        return "양수"
    else:
        return "음수"

print(check_ord_number(10))
print(check_ord_number(-20))
````

````python
# 여러개 결괏값 반환

def calc_value(a, b):
    plue = a + b
    minus = a - b
    return plue, minus # 튜플로 리턴

result1, result2 = calc_value(10, 5)
print(f"덧셈: {result1}, 뺄셈: {result2}")

result = calc_value(10, 5)
print(f"튜플:{result}")
type(result)
````

````python
# TODO: 평균 계산해서 반환하는 함수 구현
# 평균 = 총합 / 갯수
def get_average(*scores):
    if len(scores) == 0:
        return 0

    # if not scores:
    #     return

    # 합계 계산
    total = sum(scores)
    # 평균 계산 
    avg = total / len(scores)
    # 결괏값 반환
    return avg

result1 = get_average(80, 90, 100)
result2 = get_average(75, 80, 92, 100, 55)
result3 = get_average()

print(f"3과목 평균 : {result1}")
print(f"5과목 평균 : {result2}")

````

````python
# 미니 계산기 
# 사칙 연산기

def add(a, b):
    return a + b

def minus(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "0으로 나눌수 없습니다."
    return a / b

a = input("첫번째 숫자:")
b = input("두번째 숫자:")

print(f"덧셈 연산 {a} + {b}: {add(int(a), int(b))}")
print(f"뺄셈 연산 {a} - {b}: {minus(int(a), int(b))}")
print(f"곱셈 연산 {a} * {b}: {multiply(int(a), int(b))}")
print(f"나눗셈 연산 {a} / {b}: {divide(int(a), int(b))}")
````

## 변수 스코프 (유효범위)

- 전역 변수 : 함수 밖에서 선언하는 변수, 어디서나 접근이 가능
- 지역 변수 : 함수내에 선언하는 변수, 함수안에서만 접근이 가능

````python
def local():
    local_var = 1 # 지역변수
    print(local_var)

local()
# print(local_var) # 오류 발생
````

````python
# 전역변수
message = "안녕하세요. Kant몰에 오신것을 환영합니다."

def greeting():
    message = "장난감 코너는 3층에 있습니다." # 지역 변수
    print(message) # 지역변수가 우선된다.

greeting()

print(message) # 전역변수
````

````python
# 전역변수
value = "전역 데이터"

# Function ==================
def outer_function():
    value = "지역 변수 1"

    def inner_function():
        value = "지역 변수 2"
        print(value)

    inner_function()
    print(value)
# Function ==================

outer_function()
print(value) # 전역변수
````

````python
# global 키워드

counter = 0

def inc_counter():
    # counter = 0
    global counter # 전역 변수를 참조하겠다고 선언
    counter += 1

inc_counter()
inc_counter()

print(counter)
````
